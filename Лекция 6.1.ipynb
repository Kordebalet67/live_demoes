# Функции. Начало

Стандартная и простейшая инструкция

```python
print("Hello world!")
```

Оборачиваем предыдущую инструкцию в функцию.

Функция простейшая: не принимает никаких аргументов, не возвращает значения

```python
def hello():
    print("Hello world!")
```

```python
hello()
```

Усложняем.

Перепишем функцию, но с условием, что она принимает имя пользователя, и не возвращает значения

```python
def hello_user(name):
    print("Hello ", name, "!")

```

```python
hello_user("Vlad")
```

```python
my_name = "Vlad"
hello_user(my_name)
```

Сделаем функцию чуть более информативной

```python
def hello_user(name: str = 'Vlad'):
    print(f"Hello, {name}!")
```

```python
hello_user()
```

```python
hello_user('Alex')
```

```python
def hello_user(age: int = 10, name: str = 'Vlad'):
    print(f"Hello, {name}!")
```

```python
hello_user(12, 'ivan')
```

```python
hello_user(name='ivan', age=12)
```

Еще усложним. Посчитаем произведение.

Функция принимает значения двух переменных, а возвращает результат умножения

```python
def multiply(a, b):
    return a * b
```

```python
a1 = 4
b1 = 5
result = multiply(a1, b1)
print(result)
```

```python
def multiply(a: int, b: int) -> int: 
    return a * b
```

```python
multiply('1.5', 2)
```

```python
def division(a: int, b: int) -> float:
    return a / b
```

```python
division(2, 2)
```

Передадим в качестве аргумента функции коллекцию

```python
def factorial(a: list):
    res = 1
    for i in a:
        res *= i 
    return

num_list = [1, 2, 3, 4, 5, 6, 7]
result = factorial(num_list)
print(result)
```

```python
def factorial(a):
    res = 1
    for i in a:
        res *= i 
    return res

num_list = [1, 2, 3, 4, 5, 6, 7]
result = factorial(num_list)
print(result)
```

# *args

```python
def geoprogress(*args):
    result = 1
    for item in args:
        result *= item
    return result
```

```python
geoprogress(1,2,4,8,16,32,64,128,256)
```

```python
def geoprogress(args):
    result = 1
    for item in args:
        result *= item
    return result
```

```python
geoprogress(1,2,4,8,16,32,64,128,256)
```

# **kwargs

```python
def geoprogress(**kwargs):
    print(kwargs)
```

```python
geoprogress('asda': 123)
```

```python
geoprogress(123)
```

```python
geoprogress({'asda': 12321})
```

```python
geoprogress(['1','2'],[1,2])
```

```python
geoprogress(['as', 12])
```

```python
dicts = {'1': 1, '2': 2}
```

```python
geoprogress(**dicts)
```

# *args + **kwargs

```python
def simple(*args, **kwargs):
    print(f'args: {args}')
    print(f'kwargs: {kwargs}')
```

```python
simple(1,2,3,4,5, **dicts)
```

```python
def simple(arg1, arg2, *args, **kwargs):
    print(f'args: {arg1}, {arg2}')
    print(f'*args: {args}')
    print(f'**kwargs: {kwargs}')
```

```python
simple(1,2,3,4,5, **dicts)
```

# Самое полое и красивое описание функции

```python
def full(arg1: int = 1, 
         arg2: float = 1.0, 
         *args, 
         **kwargs) -> int:
    print(f'args: {arg1}, {arg2}')
    print(f'*args: {args}')
    print(f'**kwargs: {kwargs}')
    return 1
```

```python
full(1,2,3,4,5, **dicts)
```
